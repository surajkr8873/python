Use a while loop to reverse a given number.

```python
num = 789

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reverse:", reverse)
```

Use a while loop to count the number of even digits in a given number.

```python
num = 789635874123

count = 0

while num > 0:
    digit = num % 10

    if digit % 2 == 0:
        count += 1

    num = num // 10

print("Even digits:", count)
```

Create a program that keeps asking the user to enter a password until the correct password is entered.

```python
password = 1234

while True:
    user_password = int(input("Enter password: "))

    if user_password == password:
        print("Access Granted")
        break
    else:
        print("Wrong Password")
```

```python
num = 789654123 
sum = 0

while num > 0:
    last_digit = num % 10
    sum = sum + last_digit
    num = num // 10          # >> ishse decimal hat jata hai

    sum += 1

print("sum :" , sum)
```

while loop se kisi number ka factorial nikalo.
Example: 5 → 120

```python
num = int(input("enter a number : "))
fact = 1

while num > 0:
    fact = fact * num

    num -= 1
    
print("factorial :" , fact)
```
