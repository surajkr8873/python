What is operator ?

```python
# English:

# An operator is a symbol that tells Python to perform an operation on values.

# Hindi:

# Operator ek symbol hota hai jo Python ko batata hai ki values ke saath kya kaam karna hai.
```

Eg:-

```python
a = 10
b = 5

print(a + b) # 10 + 5 >> + is operator.ye a or b ko kya krna hai ye batata hai,operator.
```

some common operator.

```python
#----- (+)   → Addition
#----- (-)   → Subtraction
#----- (*)   → Multiplication
#----- (/)   → Division
#----- (% )  → Remainder
#----- (==)  → Equal to
#----- (> )  → Greater than
#----- (< )  → Less than
#----- (!=)  → Not Equal
#----- (>=)  → Greater or Equal
#----- (<=)  → Less or Equal
#----- (//)  → Floor Division
#----- (**)  → Power

# Operator = jo Python ko operation karne ka instruction deta hai.
```
