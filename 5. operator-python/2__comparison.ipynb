table of comparison operator

```python
#           Operator        	Meaning	                Example	               Result
#             ==	             Equal                  10 == 10	            True
#             !=	             Not Equal	            10 != 5	                True
#             >	                 Greater	            10 > 5	                True
#             <	                 Less	                5 < 10	                True
#             >=	             Greater/Equal	        10 >= 10	            True
#             <=	             Less/Equal   	        5 <= 10  	            True
```

SOME EXAMPLE :-

```python
a = 15
b = 5
age = 18

print(a == b) # False
print(a != b) # True
print(a > b) # True
print(a < b) # False
print(a >= b) # True
print(a <= b) # False

print(age >= 19) # False
print(age <= 19) # True
```
