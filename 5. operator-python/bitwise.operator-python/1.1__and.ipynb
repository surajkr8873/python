5 example of and (&) operator in bitwise.

```python
# Eg :- 1

a = 5    # 5 → 101
b = 3    # 3 → 011

print(a & b)    # >>> 101
                # >>> 011
                # >>> 001 >>>> answer = 1

# Eg :- 2

a = 6    # 6 → 110
b = 3    # 3 → 011

print(a & b)    # >>> 110
                # >>> 011
                # >>> 010 >>>> answer = 2

# Eg :- 3

a = 7    # 7 → 111
b = 4    # 4 → 100

print(a & b)    # >>> 111
                # >>> 100
                # >>> 100 >>>> answer = 4

# Eg :- 4

a = 10   # 10 → 1010
b = 6    # 6  → 0110

print(a & b)    # >>> 1010
                # >>> 0110
                # >>> 0010 >>>> answer = 2

# Eg :-5

a = 12   # 12 → 1100
b = 5    # 5  → 0101

print(a & b)    # >>> 1100
                # >>> 0101
                # >>> 0100 >>>> answer = 4
```
