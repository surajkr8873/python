What is  right shift operator in biitwise ?

```python
# English:

# Right Shift moves the bits to the right.

# Hindi:

# Right Shift mein binary ke bits ko right side mein shift karte hain.

# Left side mein jo bits bahar chale jaate hain, woh hat jaate hain.
```

sign of  right  shift operator in bitwis. 

```python
#  >>
```

EXAMPLE

```python
a = 10    # 10 → 1010

print(a >> 1)
```

5 example of right  shift operator in bitwise

```python
a = 10   # 10 → 1010

print(a >> 1)    # >>> 1010
                 # >>> 0101 >>>> answer = 5


a = 20   # 20 → 10100

print(a >> 2)    # >>> 10100
                 # >>> 00101 >>>> answer = 5


a = 16   # 16 → 10000

print(a >> 1)    # >>> 10000
                 # >>> 01000 >>>> answer = 8


a = 24   # 24 → 11000

print(a >> 3)    # >>> 11000
                 # >>> 00011 >>>> answer = 3


a = 32   # 32 → 100000

print(a >> 2)    # >>> 100000
                 # >>> 001000 >>>> answer = 8
```
