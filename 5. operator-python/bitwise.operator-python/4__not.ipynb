sign of not operator in bitwise

```python
# ~
```

```python
a = 5    # 5 → 101

print(~a)
```

5 example of not operator in bitwaise.

```python
a = 5    # 5 → 101

print(~a)    # >>> ~101
             # >>> -110 >>>> answer = -6


a = 6    # 6 → 110

print(~a)    # >>> ~110
             # >>> -111 >>>> answer = -7


a = 7    # 7 → 111

print(~a)    # >>> ~111
             # >>> -1000 >>>> answer = -8


a = 10   # 10 → 1010

print(~a)    # >>> ~1010
             # >>> -1011 >>>> answer = -11


a = 12   # 12 → 1100

print(~a)    # >>> ~1100
             # >>> -1101 >>>> answer = -13
```
