What is or opertor in bitwise ?

```python
# English:

# Bitwise OR compares two numbers bit by bit.

# Hindi:

# Bitwise OR do numbers ke binary bits (0 aur 1) ko ek-ek karke compare karta hai.
```

```python
# 1 | 1 → 1 >> isme jab dono me se kooi bhi bit 1 ho to ye True hota hai yani 1.
# 1 | 0 → 1
# 0 | 1 → 1
# 0 | 0 → 0 >> lekin jab bit dono 0  ho > yani False  ho to to wo 0 hota hai.
```

sign of or operator in bitwaise.

```python
# |
```

EXAMPLE

```python
a = 5    # 5 → 101
b = 3    # 3 → 011

print(a | b)    # >>> 101
                # >>> 011
                # >>> 111 >>>> answer = 7
```

5 EAMPLE of or ooperator in bitwise

```python
a = 5    # 5 → 101
b = 3    # 3 → 011

print(a | b)    # >>> 101
                # >>> 011
                # >>> 111 >>>> answer = 7


a = 6    # 6 → 110
b = 3    # 3 → 011

print(a | b)    # >>> 110
                # >>> 011
                # >>> 111 >>>> answer = 7


a = 7    # 7 → 111
b = 4    # 4 → 100

print(a | b)    # >>> 111
                # >>> 100
                # >>> 111 >>>> answer = 7


a = 10   # 10 → 1010
b = 6    # 6  → 0110

print(a | b)    # >>> 1010
                # >>> 0110
                # >>> 1110 >>>> answer = 14


a = 12   # 12 → 1100
b = 5    # 5  → 0101

print(a | b)    # >>> 1100
                # >>> 0101
                # >>> 1101 >>>> answer = 13
```
