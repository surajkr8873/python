What is left shift in bitwise?

```python
# English:

# Left Shift moves the bits to the left by a given number of positions.

# Hindi:

# Left Shift mein binary ke bits ko left side mein shift karte hain.

# Aur right side mein 0 add hota hai.
```

sign of left shift operator in bitwise.

```python
#  <<
```

EXAMPLE

```python
a = 5    # 5 → 101

print(a << 1)
```

5 example of left shift operator in bitwise.

```python
a = 5    # 5 → 101

print(a << 1)    # >>> 101
                 # >>> 1010 >>>> answer = 10


a = 6    # 6 → 110

print(a << 1)    # >>> 110
                 # >>> 1100 >>>> answer = 12


a = 3    # 3 → 11

print(a << 2)    # >>> 11
                 # >>> 1100 >>>> answer = 12


a = 7    # 7 → 111

print(a << 2)    # >>> 111
                 # >>> 11100 >>>> answer = 28


a = 10   # 10 → 1010

print(a << 3)    # >>> 1010
                 # >>> 1010000 >>>> answer = 80
```
