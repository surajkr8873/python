What is bit wise operator ?

```python
# English:

# Bitwise operators work on individual bits (0 and 1) of numbers.

# Hindi:

# Bitwise operators number ke binary form (0 aur 1) par kaam karte hain.
```

table ofbitwise operator, Python mein mainly 6 bitwise operators hain.

```python
#          Operator	                  Name	                            Simple meaning

#             &                  	  AND	                             dono bit 1 ho
#             `                    	   `	                             OR
#             ^	                      XOR	                             bits different ho
#             ~	                      NOT	                             bits ko reverse karta hai
#             <<	                  Left Shift	                     bits ko left shift karta hai
#             >>	                  Right Shift	                     bits ko right shift karta hai
```

```python
a = 5 # >>> 5 → 101,
b = 3 # >>> 3 → 011,

print(a & b)
print(a | b)
print(a ^ b)
print(~a)
print(a << 1)
print(a >> 1)
```
