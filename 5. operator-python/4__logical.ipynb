What is logical ooperator ?

```python
# English:

# Logical operators are used to combine or check multiple conditions.

# Hindi:

# Logical operators ka use 2 ya usse zyada conditions ko check/combine karne ke liye hota hai.
```

```python
# Python mein 3 main logical operators hain:


# 1. and
# 2. or
# 3. not
```

and operator >> jab dono conndition sahi ho


```python
a = 10

print(a > 5 and a < 20) # True

# a > 5   → True
# a < 20  → True

# True and True → True
```

table of and operator

```python
#                   Condition 1             	Condition                  2	Result
#                     True	                      True	                        True
#                     True	                      False	                        False
#                     False	                      True	                        False
#                     False                       False	                        False


# agar  condition dono True hai tabhi sahi hoga, warna False hoga.
```

5  example of and ooperaator.

```python
# Example 1
a = 10
print(a > 5 and a < 20)

# Example 2
a = 10
print(a > 5 and a > 20)

# Example 3
age = 20
print(age >= 18 and age <= 25)

# Example 4
a = 15
print(a % 2 == 0 and a > 10)

# Example 5
marks = 80
print(marks >= 40 and marks <= 100)
```

OR operator.

```python
# or operator >> do condition me se koi ik condition sahi ho.
```

```python
a = 10

print(a > 5 or a > 20)
```

table of or operator.

```python
#           Condition 1            Condition  2             	Result
#             True	                 True	                     True
#             True	                 False	                     True
#             False	                 True	                     True
#             False	                 False	                     False
```

5 example of or operator.

```python
# Example 1
a = 10
print(a > 5 or a > 20)

# Example 2
a = 10
print(a > 20 or a < 5)

# Example 3
age = 15
print(age < 18 or age > 60)

# Example 4
a = 15
print(a % 2 == 0 or a > 10)

# Example 5
marks = 30
print(marks >= 40 or marks == 30)
```

not operator


```python
# English:

# not operator reverses the result of a condition.

# Hindi:

# not operator True ko False aur False ko True kar deta hai.
```

```python
a = 10

print(not(a > 5)) # a bada hai 5 se lekin False output  aaya kyuki not operator use kr rhe hai.not operator condition ke result ko ulta kar deta hai.
```
