What is assignment ooperator ?

```python
# English:

# Assignment operators are used to assign  a value to a variable.

# Hindi:

# Assignment operator ka use kisi variable ke andar value rakhne ya update karne ke liye hota hai.
```

table of  assign operator

```python
#                 Operator                   	Meaning	                  Same as
#                    =	                        Assign	                   a = 10
#                    +=                      	Add	                       a = a + 5
#                    -=	                        Subtract	               a = a - 5
#                    *=	                        Multiply	               a = a * 5
#                    /=	                        Divide	                   a = a / 5
#                    %=                      	Remainder	               a = a % 5
#                   //=	                        Floor division	           a = a // 5
#                   **=  	                    Power	                   a = a ** 5
```

basiic example

```python
a = 10

print(a) # yaha pr a name ka ik box  hai jisme hum 10 ko rkh diye hai
```

```python
a = 10

# ab hum chahte hai ki a ka value hum change kare bina pichhhle walle a ko disturb kiye.

a += 5 # (yaha pr hum a me 5 add kr diye.)


print(a)
```

```python
a = 10
a = a - 5 # substraction
print(a)
```

```python
a = 10
a *= 4 # multiply
print(a)
```

```python
a = 10 
a /= 2 # division
print(a)
```

```python
a = 10
a %= 3 #mmodulus > remainder
print(a)
```

```python
a = 10 
a //= 2 # floor division >> isme decimal value  hat jayega
print(a)
```

```python
a = 10
a **= 2 # power
print(a)
```
