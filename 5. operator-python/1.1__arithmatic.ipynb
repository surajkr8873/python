What is arithmatic operator ?


```python
# English:

# Arithmetic operators are used to perform mathematical calculations on numbers.

# Hindi:

# Arithmetic operators ka use numbers ke saath calculation karne ke liye hota hai.
```

table of arithmatic  ooperator

```python
#         Operator     	Meaning	             Example	         Answer

#           +	        Addition	          10 + 5	           15
#           -	        Subtraction	          10 - 5	           5
#           *	        Multiplication	      10 * 5	           50
#           /	        Division	          10 / 5	           2.0
#           %	        Remainder	          10 % 3	           1
#          //	        Floor Division	      10 // 3	           3
#          **	        Power	              2 ** 3	           8
```

Example:-

```python
a = 10 
b = 5

print(a + b) # addition
print(a - b) # substraction
print(a * b) # multiply
print(a / b) # division
print(a % b) # modulus (remainder)
print(a // b) # floor division >> isme decimal point hat jayega
print(a ** b) # power


# these example are arithmatic ooperator
```
