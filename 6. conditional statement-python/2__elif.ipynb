What use in elif in condtional statement ?

```python
# English

# elif means "else and if".

# It is used when we want to check another condition after the previous if condition becomes False.

# Hindi

# elif ka matlab hota hai:

# "Agar pehli condition false hai, to ye wali condition check karo."
```

Elif ki jarurat kyu padti hai?

```python
# Pehli condition?
#      ↓
#    False
#      ↓
# Doosri condition?
#      ↓
#    False
#      ↓
# Teesri condition? >>>>>>> jab tak hum elif ka use krke condition check krte rhe koi end point nahi hai.

# lekin jab sari elif ka condition false ho jaye  to else ka jo andr ka code run hoga. 
```

Example:-

```python
# if condition:
#     # code

# elif condition:
#     # code

# if    → pehli condition check
# elif  → agar if False hua, doosri condition check


```

elif kaise kaam karta hai?

```python
# Python upar se neeche conditions check karta hai.

# if condition1:
#     # code

# elif condition2:
#     # code

# elif condition3:
#     # code

# else:
#     # code
```

Flow

```python
# condition1
#     ↓
# True? ─── Yes → if ka code
#     │
#     No
#     ↓
# condition2
#     ↓
# True? ─── Yes → elif ka code
#     │
#     No
#     ↓
# condition3
#     ↓
# True? ─── Yes → elif ka code
#     │
#     No
#     ↓
# else
```

5 Example of elif coondion.

```python
# 1. Marks ke according grade

marks = int(input("enter marks"))

if marks >= 90:
    print("A")
elif marks >= 80:
    print("B")
else:
    print("C")


# 2. Number Positive, Negative ya Zero

num = int(input("enter a number"))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")


# 3. Age ke according

age = int(input("enter your age"))

if age >= 18:
    print("Adult")
elif age >= 13:
    print("Teenager")
else:
    print("Child")


# 4. Temperature

temp = int(input("enter temperature"))

if temp >= 40:
    print("Very Hot")
elif temp >= 30:
    print("Hot")
elif temp >= 20:
    print("Normal")
else:
    print("Cold")


# 5. Do numbers compare karna

a = int(input("enter number"))
b = int(input("enter number"))

if a > b:
    print("a is greater")
elif a < b:
    print("b is greater")
else:
    print("Both are equal")
```
