what use in nested if in conditional statement.

```python
# English

# Nested if means an if statement inside another if statement.

# Hindi

# Nested if ka matlab hai: ek if ke andar doosra if.


# ::::::::::::::::::::::::::::::----

# Pehle ek condition check karo.
# Agar wo True hai, tab andar wali doosri condition check karo.
```

structure :-

```python
# if condition1:
#     if condition2:
#         print("condition true")


#   yaad rkhne ka technique::-

# if
#  └── if
#       └── kaam
```

1 Example in details:-


```python
age = 17
has_id = False

if age >= 18:
    # Pehle check hoga: age 18 ya usse zyada hai?
    
    if has_id == True:
        # Agar age wali condition True hai,
        # tab ye doosri condition check hogi.
        
        print("Entry Allowed")

else:
    print("not allowed")
```

5 Example of nested if:-

```python
# 1. Age + ID check

age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry Allowed")


# 2. Positive + Even check

num = 10

if num > 0:
    if num % 2 == 0:
        print("Positive Even Number")


# 3. Login check

username = "suraj"
password = "1235"

if username == "suraj":
    if password == "1234":
        print("Login Successful")
    else:
        print("jakr password check karo pahle")

# 4. Marks check

marks = 80

if marks >= 40:
    if marks >= 75:
        print("Pass with Distinction")


# 5. Number check

num = 20

if num > 0:
    if num > 10:
        print("Positive number greater than 10")
```
