If & else defition:-

if :-

```python
# if is used to check a condition.

# If the condition is True, Python runs the code inside if.

# Hindi

# if ka matlab hota hai "agar".

# Agar condition True hai → if ke andar ka code chalega.
```

else :-

```python
# English

# else means "otherwise".

# It runs when the if condition is False.

# Hindi

# else ka matlab hota hai: warna

# "Agar if ka condition sahi nahi hua to else ka code chalega."
```

Likhne ka tarika :-

```python
# if condition:
#     statement
# else:
#     statement
```

```python
age = 15

if age >= 18:
    print("You can vote")        # >> agar condition sahi hua to if ka andar wala code chlega. 
else:
    print("You cannot vote")     # >> agar if ka condition galat hua to else ka code chlega. 
```

```python
#  ik hi conditionn (if) ke andr hum kafi sare code ko run kara skte hai.else ke andr bhi.

age = 18  # >>>> yaha pr hum user se input ya value le skte hai.

if age >= 18:
    print("you can vote")
    print("you can apply for licence")
    print("you are eligible for driving")

else:
    print("you can't vote")
    print("you can't apply for licence")
    print("you are nott eligible for driving")
```

If ke andar (and) ka use.

```python
age = 20
has_id = True

if age >= 18 and has_id:
    print("Entry allowed")
else:
    print("Entry not allowed")

# age >= 18 → True
# has_id    → True  >>>> jabb condition dono True hoga tab hi if ke andar ka code  chlega  nahi to else ka code run karega
```

If ke andar (or) ka use.

```python
day = "Sunday"

if day == "Sunday" or day == "Saturday":
    print("Holiday")
else:
    print("Working day")


# Kam se kam ek condition True honi chahiye.
```

if ke andar (not) ka use.

```python
is_raining = False

if not is_raining:
    print("Go outside")
else:
    print("Stay inside")

# not True ko False aur False ko True karta hai.
```

Five Example of if else:-

```python
# 1. Even or Odd
num = int(input("enter a number"))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")


# 2. Pass or Fail
marks = int(input("enter marks"))

if marks >= 40:
    print("Pass")
else:
    print("Fail")


# 3. Adult or Not Adult
age = int(input("enter you age"))

if age >= 18:
    print("Adult")
else:
    print("Not Adult")


# 4. Positive or Negative
num = int(input("enter a number"))

if num >= 0:
    print("Positive")
else:
    print("Negative")


# 5. Greater or Smaller
a = int(input("enter a number"))
b = int(input("enter a number"))

if a > b:
    print("a is greater")
else:
    print("a is smaller or equal")
```
