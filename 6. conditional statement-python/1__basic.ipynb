```python
age = 15                   # >>> yaha pr age hai ya kuchh bhi value ho skta hai,,,ish value ko hum input me user se bhi le skte hai.

if age >= 18:              # >>> if > (agr) condition hai >>>> yaha condition hai ki age jo hai wo 18 se bada ya 18 ke barabar ho.to wo adult hoga
    print("Adult")         # >>> agar condition sahi hoga to ye print karega adult.
                               # >>> kyuki humne syntax me likha hua ki agr condition sahi hai to print kro adult.


                               # lekin jaruri nahi hai ki condition sahi hi hoga lekin agar  condition galat hua to kuchh to print ya output nikalega. 
                               # so  isiliye hum else (warna) >> krke bhi likhenge.jaise humne niche wala  line me kiya hua hai.
else:
    print("Not Adult")


# humne simple sa syntax ya code likha.>>>> ki 
# agar condition sahi hai to print kro adult.
# warnna print karo not adult.
```
