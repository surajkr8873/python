What is returns in funtion (in pyhton)

```python
# English

# return is used to send a value back from a function.

# Hindi

# return ka use function ke andar se value bahar bhejne ke liye hota hai.
```

```python
# Input → Function → Result

# return result ko function ke bahar bhej deta hai.
```

```python
def sum(a , b):  # >> a or b == (parameter)
    add = a + b
    return add   # >> 13 ko function ke bahar bhej diya.


answer = sum(5, 8)   # >> answer ke andar 13 aa gaya.

print(answer) # >> answer ko print(output) kr diya.
```

5 Example of return function inn python

```python
# multiplication

def multi(a , b):
    multiply = a * b
    return multiply

answer = multi(5 , 6)
print(answer)

# substraction

def minus(a , b):
    sub = a - b
    return sub

answer = minus( 5 , 8)
print(answer)

# addition

def add(a , b , c):
    ans = a + b + c
    return ans

answer = add(5 , 6 , 8)

print(answer)

# division

def division(a , b):
    divide = a // b # >> double slash ka matlab ki  decimal value remove ho jayega.
    return divide

answer = division(10 , 2)
print(answer)

# power

def power(a , b):
    square = a ** b
    return square

answer = power(5 , 2)
print(answer)

# 5. Greeting
def greet(name):
    return "Hello " + name

print(greet("Suraj"))
```
