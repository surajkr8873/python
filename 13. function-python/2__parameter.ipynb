What is parameter in funcion in python?

```python
# English

# A parameter is a variable written inside the function's brackets () that receives a value when the function is called.

# Hindi

# Parameter ek variable hota hai jo function banate time () ke andar likhte hain. Function call hone par usme value aati hai.
```

Example:-

```python
def func_name(name):   # >>> name = parameter.Ye ek khaali box jaisa hai.
    print("hello" , name)


func_name("suraj")     # >>>  "Suraj" = argument.Ye value name ke box mein chali gayi.


# Parameter = name
# Argument  = "Suraj"
```

5 Example of parameter  funnction in python.

```python
# 1
def greet(name):
    print("Hello", name)

greet("Suraj")


# 2
def add(a, b):
    print(a + b)

add(10, 20)


# 3
def square(num):
    print(num * num)

square(5)


# 4
def welcome(city):
    print("Welcome to", city)

welcome("Patna")


# 5
def student(name, age):
    print(name, age)

student("Suraj", 18)
```

```python
# Function banate time → Parameter
# Function call karte time → Argument


def add(a, b):      # a, b = parameters
    print(a + b)

add(10, 20)         # 10, 20 = arguments >> jab code run hota hai tab arguments parameter me chla me jata hai.
```
