What is default argument function in pyhton.

```python
# Agar function call karte time hum argument na dein, to Python pehle se diya hua default value use kar lega.


# English

# A default argument is a parameter that already has a value.

# Hindi

# Function ke parameter ko pehle se ek value de dena = Default Argument.
```

basic syntax:-

```python
def function_name(name="Suraj"):       
    print(name)

function_name()


# name="Suraj" → default argument
# functio_name() → humne koi value nahi di
# Python automatically "Suraj" use karega.
```

agar value de diya jaye arguments me 

```python
def greet(name="Suraj"):
    print("Hello", name)

greet("Rahul")

# Jab hum value dete hain, default value replace ho jaati hai.
```

5 examplpe of default aaguments function in python

```python
# 1. Name
def your_name(name="Suraj"):
    print("Hello", name)

your_name()

# 2. Age
def age(age=18):
    print("Age =", age)

age()

# 3. Addition
def add(a, b=10):
    print(a + b)

add(5)

# 4. City
def city(city="Siwan"):
    print("City =", city)

city()

# 5. Welcome
def welcome(name="User"):
    print("Welcome", name)

welcome()
welcome("Suraj")
```
