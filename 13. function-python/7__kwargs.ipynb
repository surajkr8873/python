What is **kwarg function in python

```python
# Function mein jitne chahe key=value arguments bhej sakte ho.
```

*args mein values aati hain:

```python
# add(10, 20, 30)
```

Lekin **kwargs mein naam ke saath value aati hai:

```python
# student(name="Suraj", age=18, city="Siwan")
```

Basic syntax

```python
def function_name(**kwargs):
    print(kwargs)
```

Example:

```python
def student(**kwargs): # >> kwargs ke jagah pr kuchh bhi likh skte hai lekin kuchh likhne se phle  ** denna jaruri ai
    print(kwargs)
    print(type(kwargs))

student(name="Suraj", age=18, city="patna") # >> ye hamesha dictionary ke typr me aayegga
```

simple roop se smjhte hai

```python
# socho tmhare pass ik bag hai (**kwargs) > Is bag mein har cheez ka naam + value hai:

# name → Suraj
# age  → 18
# city → patna


#    { "name": "Suraj",
#     "age": 18,
#     "city": "patna" }
```

for loop ke saath

```python
def student(**kwargs):

    for key, value in kwargs.items():
        print(key,"=" , value)

student(name="Suraj", age=18, city="patna")



# name = Suraj
# age = 18
# city = patna


# name , age , city >>> keys
# suraj , 18 , patna >> values
```

5 example.

```python
# 1. Student details
def student(**kwargs):
    print(kwargs)

student(name="Suraj", age=18, city="Siwan")


# 2. Employee details
def employee(**kwargs):
    for key, value in kwargs.items():
        print(key, "=", value)

employee(name="Rahul", salary=20000)


# 3. Multiple details
def details(**kwargs):
    print(kwargs)

details(name="Aman", age=20, course="Python")


# 4. Count key-value pairs
def count(**kwargs):
    print(len(kwargs))

count(name="Suraj", age=18, city="Siwan")


# 5. Check type
def show(**kwargs):
    print(type(kwargs))

show(name="Suraj", age=18)
```

```python
# *args = values 
# **kwargs = key + value
```
