What is function ?

```python
# English

# A function is a block of code that performs a specific task.
# We write it once and can use it many times.

# Hindi

# Function code ka ek chhota block hota hai jo ek particular kaam karta hai.
# Ek baar function bana do, phir usko baar-baar call kar sakte ho.


# Input → Machine → Output >>>> Python mein function bhi aisa hi ho sakta hai.
```

Basic syntax

```python
# def function_name():   >>> (function_name) ko humm ik variable ke jaisa trate  karte ha. >>>>>> function is (def) >> funtion ke baad hum next line tab space dekr likhte hai code ko
#     # code           


# >>> 1.- def → function banane ke liye
# >>> 2.- function_name → function ka naam
# >>> 3.- () → parameters/arguments ke liye
# >>> 4.- : → function ka block start
# >>> > - andar ka code → jo kaam function karega
```

Example:-

```python
def name():           # >> name naam ka function bana rahe hain.
    print("suraj")    # >> Function ke andar ka kaam hai.
name()                # >> Function ko call kiya.
                      # >> Ab function ke andar wala code chalega.

# output >> suraj
```

5 simple example of function in python.

```python
# 1. Greeting
def hello():
    print("Hello")

hello()


# 2. Name print
def name():
    print("Suraj")

name()


# 3. Addition
def add(a , b):
    sum = a + b
    print(sum)

add(5 , 6)


# 4. Multiplication
def multiply(a , b , c):
    multiplication = a*b*c
    print(multiplication)

multiply(5 , 6 , 8)



# 5. Message
def message():
    print("I am learning Python")

message()
```
