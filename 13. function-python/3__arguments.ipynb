What is arguments in function in python ?

```python
# English

# Arguments are the actual values we pass to a function when we call it.

# Hindi

# Function ko call karte time () ke andar jo actual value dete hain, usko argument kehte hain.
```

Example

```python
def greet(name):
    print("Hello", name)

greet("Suraj")


# name   → Parameter
# "Suraj" → Argument
```

Simmple way to understand arguments in function.

```python
# jab hum function ko call krte time jo value likhte hai usko hum arguments khte hai.
```

5 Example of arguments  function in python

```python
# 1
def greet(name):
    print("Hello", name)

greet("Suraj")          # "Suraj" = argument


# 2
def add(a, b):
    print(a + b)

add(10, 20)             # 10, 20 = arguments


# 3
def square(num):
    print(num * num)

square(5)               # 5 = argument


# 4
def welcome(city):
    print("Welcome to", city)

welcome("Patna")        # "Patna" = argument


# 5
def student(name, age):
    print(name, age)

student("Suraj", 18)    # "Suraj", 18 = arguments
```

```python
# Function banate time:
# def add(a, b):
#         ↑  ↑
#     Parameters

# Function call karte time:
# add(10, 20)
#     ↑   ↑
#   Arguments
```
