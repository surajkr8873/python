for loop in function *arg in python.

```python
# for loop ke saath

# *args ke andar multiple values hoti hain, isliye hum loop laga sakte hain:

def numbers(*args):

    for i in args:
        print(i)

numbers(10, 20, 30, 40)
```

*arg se  sum

```python
def add(*args):

    total = 0

    for i in args:
        total = total + i

    print(total)

add(10, 20, 30, 40)
```

sub in *arg

```python
def sub(*arg):

    total = arg[0]

    for i in arg[1:]:
        total = total - i

    print(total)

sub(10 , 5 )
```

5 example of *arg function in python

```python
# 1. Multiple names
def names(*args):
    print(args)

names("Suraj", "Rahul", "Aman")


# 2. Multiple numbers
def numbers(*args):
    for i in args:
        print(i)

numbers(10, 20, 30, 40)


# 3. Sum
def add(*args):
    total = 0

    for i in args:
        total = total + i

    print(total)

add(5, 10, 15)


# 4. Count arguments
def count(*args):
    print(len(args))
    print(type(args))

count(10, 20, 30, 40, 50)


# 5. Check type
def show(*args):
    print(type(args))

show(10, 20, 30)
```
