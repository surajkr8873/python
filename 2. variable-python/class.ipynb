WHAT IS VARIABLE

```python
# English:
# A variable is a name used to store a value.

# Hindi:
# variable ik naam wala box hai jisme koi value store hota hai.

# Eg :- name = "suraj kumar" >> name = variable , and name me ik value store hai suraj kumar se. to hum keh skte hai ki variable me value store hote hota hai.
#       age = 17 >> age is variable and 17 is value.

# NOTE:

# 1. value jab hum A to Z tak add krte hai to hmesa quote ke andar add krenge. ("")('')('''''')
# single quote('')
# double quote("")
# triple quote('''''')

# 2. jab hum kisi value ko number me ( 0 , 9) tak store krenge to koi quote ka use NAHI krenge.
```
