What  is dictionary ?

```python
# English 

# A dictionary is a collection of data stored in key-value pairs.

# Hindi 

# Dictionary ek collection hai jisme data (key : value) ke pair mein store hota hai.

# "suraj" : "8969532474" >>> suraj is key 
                        # >> 8969532474 is value.
```

sign of dictionary.

```python
{} 
```

```python
# dictionary is in ordered or its also a mutable

# ordered >> Matlab jis order mein items add karte ho, woh order preserve hota hai.
# mutable >> iska matlab value  ko change kar skte ho

# Eg:- 
student = {"name": "Rahul"}

student["name"] = "Aman"

print(student)

student = {"name" : "suraj" , "age" : 18 , "email" : "suraj@gmail.com" }
# >> isme key ko nahi change kiya jaa skta hai,isme sirf value ko change kar skte hai.
print(student)
```

type checking of dictionary


```python
student = {"name" : "suraj" , "age" : 18 , "email" : "suraj@gmail.com" }
print(type(student))  # >> jab hum kisi variable ka type check krte hai to paranthesis ke andar type ka use krte hai. :- print = syntax
                                                                                                               # type = type check karna ho to.
                                                                                                               #      = kis variable ka type check krna hai.
```

5 example of dictionary.

```python
# Example 1: Simple Dictionary
student = {
    "name": "Rahul",
    "age": 18
}

print(student)


# Example 2: Value access karna
student = {
    "name": "Rahul",
    "age": 18
}

print(student["name"])
print(student["age"])


# Example 3: Value change karna
student = {
    "name": "Rahul",
    "age": 18
}

student["age"] = 19

print(student)


# Example 4: New value add karna
student = {
    "name": "Rahul",
    "age": 18
}

student["city"] = "Patna"

print(student)


# Example 5: Type check
student = {
    "name": "Rahul",
    "age": 18
}

print(type(student))
```
