WHAT IS FLOAT DATATYPE ?

```python
# English:
# (float) stands for Floating Point Number. It is used to store numbers with decimal points.

# Hindi:
# (float) ka use decimal wale numbers ko store karne ke liye hota hai.

#  float value bhi negative me hho skta hai.

# Eg :- 25.2 , 58.6 , 78.2 , 55.5, .........E.tc
```

```python
price = 25.99
fee = 37.5
temperature = 24.6

# upr jitne bhi variable ke andar value hai wo sab float datatype me hai >> kyuki hum jante hai ki float data type decimal value me hote hain.
```

type check of float datatype.

```python
price = 25.99
print(type(price)) # jab hum kisi variable ka type check krte hai to paranthesis ke andar type ka use krte hai. :- print = syntax
                                                                                                               # type = type check karna ho to.
                                                                                                               #      = jis variable ka type check karna hai. 
```

float type check (example)

```python
price = 25.99
fee = 37.5
temperature = -24.6

print(type(price))
print(type(fee))
print(type(temperature))
```

```python
# in sab ka value display karna hai to type() ka use nh krenge.

price = 25.99
fee = 37.5
temperature = -24.6

print(price)
print(fee)
print(temperature)
```
