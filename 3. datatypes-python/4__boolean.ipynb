What is boolean datattpe ?

```python
# English:
# Boolean is a data type that has only two values: True or False.

# Hindi:
# Boolean ek data type hai jisme sirf 2 values hoti hain: True or False.
```

```python
# boolean ik aisa datatype hai jo hame sahi or galat ke form me information deta hai.
```

```python
student = True
raining = False
print(type(student))
print(type(raining))   # jab hum kisi variable ka type check krte hai to paranthesis ke andar type ka use krte hai. :- print = syntax
                                                                                                               # type = type check karna ho to.
                                                                                                               #      = jis variable ka type check karna hai.
```

Iska use hum kisi 2 value ko compare krne ke liye bhi use karte hai.

```python
# Eg :-

a = 10 # integer
b = 5  # integer

print(a > b)
print(type(a and b))
```
