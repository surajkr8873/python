What is string ?

```python
# English:
# A string is a collection of characters/text. In Python, strings are written inside quotes (" " or ' ').

# Hindi:
# String ka use text ya characters ko store karne ke liye hota hai. String ko hum quotes ke andar likhte hain.
```

```python
name = "Suraj"
city = "Patna"
message = "Hello" 

# upar jitne bhi variable ke andar jo value hai wo string hai.
```

string rule.>

```python
# 1. string ko hmesa quote ke andar likha jata hai.

# ('') >> single quote.
# ("") >> double quote.
# ('''''') >> triple quote

# -------------------------------------------------------------------------------------------------------------------------------------------------------

# string ko hmesa capital(A) to (Z) tak or small(a) to (z) tak likha jata hai - (quote ke andar)
```

type check of string.

```python
name = "Suraj"
print(type(name)) # jab hum kisi variable ka type check krte hai to paranthesis ke andar type ka use krte hai. :- print = syntax
                                                                                                               # type = type check karna ho to.
                                                                                                               #      = jis variable ka type check karna hai. 
```

example of type check (string)

```python
name = "Suraj"
city = "Patna"
message = "Hello" 

print(type(name))
print(type(city))
print(type(message))
```
