What is tuple ?

```python
# English

# A tuple is an ordered collection of multiple values that cannot be changed after it is created.

# Hindi 

# Tuple ek ordered collection hai jisme hum multiple values store kar sakte hain, lekin tuple banne ke baad uski values ko change nahi kar sakte.
```

```python
# Tuple bhi ek container hai jisme hum multiple values store kar sakte hain.
```

sign of tuple

```python
tup = (10, 20, 30)
print(tup)
```

```python
# List → Mutable
# Tuple → Immutable

tup = (10, 20, 30)

tup[1] = 200 # >> tuple ko channge nahi kar skte hai isiliye output :-

# Error
```

```python
# Jis order mein values daalte ho, wahi order maintain hota hai.

tup = (10, 20, 30, 40)

print(tup)
```

5 Example of  tuple.

```python
# Example 1: Simple tuple
tup = (10, 20, 30)
print(tup)


# Example 2: Indexing
tup = (10, 20, 30, 40)
print(tup[0])
print(tup[2])


# Example 3: Type check
tup = (10, 20, 30)
print(type(tup))


# Example 4: Different datatypes
tup = (10, 2.5, "Hello", True)
print(tup)


# Example 5: Immutable
tup = (10, 20, 30)

# tup[0] = 100     # ❌ Error
print(tup)
```
