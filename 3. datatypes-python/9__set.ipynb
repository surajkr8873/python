5 Example of set.

```python
# Example 1: Simple Set
s = {10, 20, 30, 40}

print(s)


# Example 2: Duplicate values
s = {10, 20, 10, 30, 20}

print(s)
# Duplicate automatically remove ho jayenge


# Example 3: Type check
s = {10, 20, 30}

print(type(s))
# <class 'set'>


# Example 4: Add value
s = {10, 20, 30}

s.add(40)

print(s)


# Example 5: Remove value
s = {10, 20, 30}

s.remove(20)

print(s)
```
