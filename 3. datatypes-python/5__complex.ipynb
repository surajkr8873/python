What is complex number?

```python
# A complex number is a number that combines a real number and an imaginary number.

# Eg :-
# 1 >>  5 + 8j >>> 5 is number,,,8j is imaginary number.
# code me hum imaginary number ko (j) se deenote krte hai.
```
