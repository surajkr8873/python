```python
# list ko hum aise smjh skte hai ki hmare pass ik bag hai jisme hum bahut sa saman rkh skte hai.
# to list bhi bag ki tarah hi hai lekkin list ke andar hum bahut sare datatype ko rkhte hai 

data = [10, 2.5, "Hello", True]

# 10       → int
# 2.5      → float
# "Hello"  → string
# True     → bool
```

```python
# list me har item ka ik address/number hota hai.jise indexing khte hai.

# python me counting 0 se suru hoti hai.

numbers = [10, 20, 30, 40, 50]

print(numbers[0])   # 10
print(numbers[1])   # 20
print(numbers[2])   # 30
print(numbers[3])   # 40
print(numbers[4])   # 50

# Index:     0    1    2    3    4
# Value:    10   20   30   40   50
```

```python
# list ki value ko change kar skte hai.list is mutable >> mutable means ki ushe change kar skte hai hai.

numbers = [10, 20, 30, 40]

numbers[1] = 200

print(numbers)
```

Some example of list

```python
numbers = [10, 20, 30]

print(numbers)       # [10, 20, 30]

print(numbers[0])    # 10

numbers[1] = 200     # 20 ko 200 kar diya

print(numbers)
```
