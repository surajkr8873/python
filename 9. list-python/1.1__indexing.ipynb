What is indexinng in list datatype ?

```python
# English

# Indexing means accessing a particular element of a list using its position number.

# Hindi

# Indexing ka matlab hai list ke andar kisi particular value ko uski position number se nikalna.
```

```python
numbers = [10, 20, 30, 40, 50]

#      Value:    10    20    30    40    50
#      Index:    0     1     2     3     4

# indexing hamesha ZERO (0) se strt hoti hai.
```

list[index]

Kisi value ko nikalne ka syntax:

```python
# list[index]
```

Example

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[0])                          #         numbers[0]
                                           #              ↓
                                           #         index 0
                                           #             ↓
                                           #            10
```

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[0])
print(numbers[1])
print(numbers[2])
print(numbers[3])
print(numbers[4])

# ishi ko agar minus indexinng me likha jata hai too usko kuchhh ish type se likhte hai

print("- indexing start")

print(numbers[-1]) # >> ishse ulta index chlti hai jo last index bota hai wo phle hota hai 

# + wala me right to left chal rha tha 
# - wala me riight to left chlaga

print(numbers[-2])
print(numbers[-3])
print(numbers[-4])
print(numbers[-5])
```

Indexing se value CHANGE bhi kar sakte hain

Ye bahut important hai because list mutable hoti hai.

```python
numbers = [10, 20, 30, 40]

numbers[1] = 200

print(numbers)
```

Index out of range

```python
# agar list ka index 4 hai or hum print 5 list ka index krenge to error aayega.

numbers = [10, 20, 30, 40, 50]
print(numbers[5])

#   ERROR
```

Nested List mein indexing

```python
# List ke andar list  kko nested list khte hai.
numbers = [[10, 20], [30, 40]]

# Index 0 → [10, 20]
# Index 1 → [30, 40]

print(numbers[0])
print(numbers[1])

print(numbers[0][0])
print(numbers[0][1])
print(numbers[0][0])
print(numbers[0][1])
```

print(numbers[0][1])

```python
# 1. First element

numbers = [10, 20, 30, 40, 50]
print(numbers[0])


# 2. Third element

numbers = [10, 20, 30, 40, 50]
print(numbers[2])


# 3. Last element

numbers = [10, 20, 30, 40, 50]
print(numbers[-1])


# 4. Change element

numbers = [10, 20, 30, 40, 50]
numbers[2] = 300
print(numbers)


# 5. Nested list

numbers = [[10, 20], [30, 40]]
print(numbers[1][0])
```
