What is list comprehension ?

```python
# English

# List comprehension is a short and easy way to create a new list.

# Hindi

# Normal for loop se list banane ke bajay hum ek hi line mein list create kar sakte hain.
```

EG :-

```python
numbers = []

for i in range(1, 6):
    numbers.append(i)

print(numbers)
```

List Comprehension way

```python
numbers = [i for i in range(1, 6)]

print(numbers)
```

Basic Syntax

```python
squares = [i * i for i in range(1, 6)]

print(squares)
```

5 Easy Examples

```python
# 1. Numbers ki list

a = [i for i in range(1, 6)]
print(a)
# [1, 2, 3, 4, 5]


# 2. Square

a = [i * i for i in range(1, 6)]
print(a)
# [1, 4, 9, 16, 25]


# 3. Even numbers

a = [i for i in range(1, 11) if i % 2 == 0]
print(a)
# [2, 4, 6, 8, 10]


# 4. Odd numbers

a = [i for i in range(1, 11) if i % 2 != 0]
print(a)
# [1, 3, 5, 7, 9]


# 5. Number ko double karna

a = [i * 2 for i in range(1, 6)]
print(a)
# [2, 4, 6, 8, 10]
```

```python
# [expression for item in iterable]

# Aur condition ke saath: 

# [expression for item in iterable if condition]

# List Comprehension = for loop ko short form mein use karke list banana.
```
