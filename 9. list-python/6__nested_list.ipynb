What is nested list iin list datatypes.

```python
# English

# A nested list is a list inside another list.

# Hindi

# Ek list ke andar agar doosri list ho, to usse Nested List kehte hain.
```

Eg:-

```python
numbers = [[10, 20, 30], [40, 50, 60]]

print(numbers)


#                                 numbers
#                                    ↓
#                      [ [10, 20, 30], [40, 50, 60] ]
#                             ↑              ↑
#                           List 1         List 2
```

```python
numbers = [[10, 20, 30], [40, 50, 60]]

print(numbers[0])
print(numbers[0][1])

#   numbers[0]     → [10, 20, 30]
#   numbers[0][1]  → 20
```

5 Example of nested list.

```python
# 1

a = [[1, 2], [3, 4]]
print(a[0][1])
# 2


# 2

a = [[10, 20], [30, 40]]
print(a[1][0])
# 30


# 3

a = [[1, 2, 3], [4, 5, 6]]
print(a[1])
# [4, 5, 6]


# 4

a = [[10, 20], [30, 40]]
a[0][1] = 200
print(a)
# [[10, 200], [30, 40]]


# 5

a = [[1, 2, 3], [4, 5, 6]]
print(a[0][2])
# 3
```

```python
#  Pehla index = kaunsi inner list
#  Dusra index = inner list ke andar kaunsa element.
```
