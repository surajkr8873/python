What is list datatypes in python ?

```python
# English

# A list is a collection used to store multiple values in a single variable.

# Hindi

# List ek aisa box hai jisme hum ek hi variable ke andar multiple values store kar sakte hain.

# list is mutable and ordered value.

# mutable >> value ko change kiya jaa skta hai.
# ordered >> value ko order me print kr skta hai.
```

Example :-

```python
numbers = [10, 20, 30, 40]

# Yahan numbers ek list hai aur uske andar 4 values hain.
```

1. List kaise banate hain?

List ke liye square brackets [ ] use hote hain.

```python
numbers = [10, 20, 30, 40]

print(numbers)
```

2. List mein different datatypes

Ek list mein alag-alag datatype bhi rakh sakte hain.

```python
data = [10, 2.5, "Suraj", True]
print(data)


#            10       → int
#            2.5      → float
#            "Suraj"  → string
#            True     → bool

# list me hum kisi bhi datatypes koo rakh skte hai
```

3. List Ordered hoti hai

List values ko order mein store karti hai.

```python
a = [10, 20, 30, 40]

# 10 → 20 → 30 → 40

print(a)
```

5 Example oof list

```python
# 1. Create a list

numbers = [10, 20, 30, 40]
print(numbers)


# 2. Access using index

numbers = [10, 20, 30, 40]
print(numbers[0])
print(numbers[2])


# 3. Change a value

numbers = [10, 20, 30]
numbers[1] = 200
print(numbers)


# 4. Add a value

numbers = [10, 20, 30]
numbers.append(40)
print(numbers)


# 5. Loop through list

numbers = [10, 20, 30, 40]

for i in numbers:
    print(i)
```

```python
#                   List
#                    ↓
#           Multiple values store
#                    ↓
#           [ ] square brackets
#                    ↓
#                 Ordered
#                    ↓
#            Indexing starts from 0
#                    ↓
#            Mutable = values change kar sakte hain
#                    ↓
#              Duplicates allowed
#                    ↓
#             type() → list
```
