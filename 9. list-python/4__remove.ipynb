Remove.

```python
# list se kisi particular value ko delete/hatana ho to remove() use karte hain.

# Eg :-

# list.remove(value)
```

EXAMPLE :-

```python
numbers = [10, 20, 30, 40, 50]

numbers.remove(30)

print(numbers)
```

5 Examples

```python
# 1
a = [10, 20, 30]
a.remove(20)
print(a)
# [10, 30]


# 2
a = ["apple", "mango", "banana"]
a.remove("mango")
print(a)
# ['apple', 'banana']


# 3
a = [1, 2, 3, 4, 5]
a.remove(1)
print(a)
# [2, 3, 4, 5]


# 4
a = [10, 20, 10, 30]
a.remove(10)
print(a)
# [20, 10, 30]
# Pehla 10 remove hua


# 5
a = [5, 10, 15]
a.remove(15)
print(a)
# [5, 10]
```
