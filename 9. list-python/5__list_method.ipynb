What is list method in list datatype .

```python
# List methods woh built-in methods hain jo list ke saath kaam karne ke liye use hote hain. Jaise add, remove, change, arrange etc.
```

Important List Methods

```python
#                 Method	                                     Kaam

# -----------------------------------------------------------------------------------------------------------------------------------

#                 append()              |                   End mein value add
#                 insert()	            |                   Kisi position/index par value add
#                 remove()              |                	Particular value remove
#                 pop()	                |                   Index ki value remove
#                 clear()	            |                   Puri list empty
#                 index()	            |                   Value ka index find
#                 count()	            |                   Value kitni baar hai
#                 sort()	            |                   Ascending order mein arrange
#                 reverse()             |                	List ko reverse
#                 copy()	            |                   List ki copy banana
#                 extend()	            |                   Do lists ko jodna
```

```python
#   Add:      |      append(), insert(), extend()

#   Remove:   |      remove(), pop(), clear()

#   Find:     |      index(), count()

#   Arrange:  |      sort(), reverse()

#   Copy:     |      copy()
```

sab list method ka Example.

```python
# 1. append() → end mein value add

a = [10, 20, 30]
a.append(40)
print(a)
# [10, 20, 30, 40]


# 2. insert() → kisi index par value add

a = [10, 20, 40]
a.insert(2, 30)
print(a)
# [10, 20, 30, 40]


# 3. remove() → particular value remove

a = [10, 20, 30]
a.remove(20)
print(a)
# [10, 30]


# 4. pop() → index ki value remove

a = [10, 20, 30]
a.pop(1)
print(a)
# [10, 30]


# 5. clear() → puri list empty

a = [10, 20, 30]
a.clear()
print(a)
# []


# 6. index() → value ka index find

a = [10, 20, 30]
print(a.index(20))
# 1


# 7. count() → value kitni baar hai

a = [10, 20, 10, 30, 10]
print(a.count(10))
# 3


# 8. sort() → ascending order

a = [30, 10, 20]
a.sort()
print(a)
# [10, 20, 30]


# 9. reverse() → list ko reverse

a = [10, 20, 30]
a.reverse()
print(a)
# [30, 20, 10]


# 10. copy() → list ki copy

a = [10, 20, 30]
b = a.copy()
print(b)
# [10, 20, 30]


# 11. extend() → doosri list ki values add

a = [10, 20]
b = [30, 40]
a.extend(b)
print(a)
# [10, 20, 30, 40]
```
