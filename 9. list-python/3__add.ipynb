5 Example of add in  list datatype

```python
# 1

a = [1, 2, 3]
a.append(4)
print(a)

# [1, 2, 3, 4]


# 2

a = ["Apple", "Mango"]
a.append("Banana")
print(a)

# ['Apple', 'Mango', 'Banana']


# 3

a = []
a.append(10)
print(a)

# [10]


# 4

a = [10, 20]
a.append(30)
a.append(40)
print(a)

# [10, 20, 30, 40]


# 5

a = [1, 2, 3]
a.append([4, 5])
print(a)

# [1, 2, 3, [4, 5]]
```
