What is slicing in list datatypes ?

```python
# English

# Slicing is used to get a part of a list.

# Hindi

# Slicing ka use list me se kuch values ka portion nikalne ke liye hota hai.

#                              list[start : stop]


#                         start → kahan se start karna hai
#                         stop → kahan tak jana hai
#                         stop wala index include nahi hota 
```

Example

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])

#          Index:    0    1    2    3    4
#          Value:   10   20   30   40   50

# >>>>>>>>      index 1 se start
# >>>>>>>>      index 4 se pehle stop
```

5 example of slicing of list datatypes .

```python
# 1

a = [10, 20, 30, 40, 50]
print(a[0:3])

# Output: [10, 20, 30]


# 2

a = [10, 20, 30, 40, 50]
print(a[2:5])

# Output: [30, 40, 50]


# 3

a = [10, 20, 30, 40, 50]
print(a[:3])

# Output: [10, 20, 30]
# start nahi diya → beginning se


# 4

a = [10, 20, 30, 40, 50]
print(a[2:])

# Output: [30, 40, 50]
# stop nahi diya → end tak


# 5

a = [10, 20, 30, 40, 50]
print(a[-4:-1])

# Output: [20, 30, 40]
```

Step ke saath slicing.  >>>>  [list[start : stop : step]]

```python
a = [10, 20, 30, 40, 50, 60]

print(a[0:6:2])

#       list[start : stop : step]
# Start included  | Stop excluded  | Step = kitna jump karna hai.
```
