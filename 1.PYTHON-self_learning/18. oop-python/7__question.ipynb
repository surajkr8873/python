```python
# Python OOPs - Interview Specific Questions

# 1. Does Python support method overloading?
# Answer: No, Python does not support traditional method overloading.
# We can achieve similar behavior using default arguments or *args.

# 2. What is the difference between == and is?
# Answer:
# == compares the values of two objects.
# is checks whether two variables refer to the same object in memory.

# 3. Can we create an object of an abstract class?
# Answer: No, we cannot directly create an object of an abstract class.

# 4. What happens if a child class does NOT implement an abstract method?
# Answer: If the child class still has abstract methods,
# Python raises a TypeError when we try to create its object.

# 5. Is Python a pure OOPs language?
# Answer: No, Python supports OOP, procedural programming,
# and functional programming.

# 6. Why is self not a keyword?
# Answer: self is a convention, not a keyword.
# We can technically rename it, but using self is recommended.

# 7. What happens if you forget to use self?
# Answer: If we assign a variable inside a method without self,
# it becomes a local variable instead of an instance attribute.

# 8. What is name mangling?
# Answer: Private variables like __x are internally renamed
# using the format _ClassName__x.

# 9. Can multiple inheritance cause ambiguity?
# Answer: Yes, multiple inheritance can cause ambiguity.
# Python resolves method lookup using MRO (Method Resolution Order),
# which follows C3 Linearization.

# 10. Is function overloading possible using default arguments?
# Answer: Python can simulate function overloading using
# default arguments, but it does not support traditional
# function overloading based only on parameter signatures.
```
