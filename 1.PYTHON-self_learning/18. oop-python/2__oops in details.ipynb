What is oops & why it exits ?

```python
# oops(object oriented proogramming) is a way of writing program by combining data and behavior together using classes and objects. 

# focuses on real-world entities. 
# data + method are tightly coupled. improves reusability, security, and scalibility. 
```

Functional programming [ FP ]

```python
# programs are writen using functions only, and data is passed between them. 

#  -- focuses on function. 
#  -- data iis mostly separated. 
#  -- less suited for large,real-world system. 
```

in oops [ code ]

```python
# think  of a student as a sngle intity. 
# -student has data(name, marks)
# -students knows what to do(calculate_grade)

#  - data + behavior

# code

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def calculate_grade(self):
        return "A" if self.marks > 80 else "B"

Student_1 = Student("Suraj Kumar", 88)
grade = Student_1.calculate_grade()
print(grade)
Student_1.name = "Suraj Kumar"
print(Student_1.name)
```

```python
# constructor >>> constructs objects

# A constructor is a  special method that is automatically called when an object is created.

# it iis initialize object data
# __init__

# eg:-

class Student:
    def __init__(self, name, roll_no, marks):       # __init__ >>>>> constructor
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

# s1 = Student()

# jab aap student class ko call  kr rhe hos1 me to aap Student ko nahi __init__ ko call kkar rhe hote ho.

# jo self me construct kar rhe the wo yaha pr call hoga

s1 = Student("suraj kumar", 6, 58)
print("Name :", s1.name)
print("Roll_no :", s1.roll_no)
print("Marks :", s1.marks)

# agar aap  kisi value ko ya name ko change krna chahte hai to wapas se instance variable ka use kr skte hai kuchh ish tarah.

s1.name = "Monu kumar"
s1.roll_no = 5
s1.marks = 99

print("================= instance variable ====================")

print("Name :", s1.name)
print("Roll_no :", s1.roll_no)
print("Marks :", s1.marks)
```

super() method :-

```python
# super() is used to call method or constructor of the parent class from the  child  class. 
#  - avoid code dulpication
#  - ensure parent initiallization
#  - works correctly with multiple inheritance

class Company:
    def __init__(self, comapnyName):
        self.companyName = comapnyName
    def address(self):
        print("location : banglore")
    def getname(self):
        return self.companyName

class Employee(Company):
    def __init__(self, name, id, companyName):
        self.name = name
        self.id = id
        super().__init__(companyName)

    def work(self):
        print("work : software developer")

    def getname(self):
        print(super().getname())
        return self.name

emp1 = Employee("Suraj kumar,", "12345," , "Google")
emp1.work()
emp1.address()
print(emp1.name, emp1.id, emp1.companyName)
print(emp1.getname())
```

MRO (method resolution order):-

```python
# MRO (method resolution order) is the order in which python searches for a method when a method is called on object.

# used when :
    # multiple inheritance
    # hybrid inheritance
    # with super()

# python uses :
    # C3 linearization algorithm

# Ensure :
    # left to right order
    # no parent is skipped
    # consistent heirachy

# Eg :-

class Father:
    def drive(self):
        print("father drive")

class Mother:
    def drive(self):
        print("mother drive")

class Child(Mother, Father):
    pass

ch1 = Child()
ch1.drive()

# isme father bhi drive kar skte hai or mother bhi to child kisko print krega jo phle inheri kiya ho child usko print krega. 
```
