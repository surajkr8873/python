What is inheritance in oop ?

```python
# inheritance is a mechanism where a child acquires properties and methods of a parent class. 
#           - code reusability
#           - better organisation
#           - less duplication
```

basic Example :-

```python
# normal class code :-

class Bird:
    def eat(self):
        print("Bird is eating")

    def fly(self):
        print("Bird is flying")

class dog:
    def eat(self):
        print("dog is eating")

    def bark(self):
        print("dog is barking")

class fish:
    def eat(self):
        print("fish is eating")

    def swim(self):
        print("fish is swimming")


# ab isko jo print karna hai usko call krke print kar sktte hai lekin isme jab hum pata hai ki sab eat kar rhe hai to phir code me redundancy badhane se code bahut bada ho jata hai isiliye inheritance oops ka use hotta hai ki kisi class ko ye inherit kar skte hai

# or agar aise code ko likhenege to baad me jab change karna hoga to ik ik code ko change krna pdega.isliye hum kuchh ish tarah likhenge:

class Animal:
    def eat(self):
        print("eating")

    # ab sabko alag alag eat ka function banane ka jarurat nhi hai hum sirf ab class banayenge or use sirf uske kaam ko add krenge eating chhodkr or Animal class ko inherit karenge.

class Fish(Animal):       # ish tarah Animal class ko sabme inherit krenge jitne bhi class banenge sabme.abs
    def swim(self):
        print("fish is swimming") 

class Dog(Animal):
    def bark(self):
        print("dog is barking")

class Bird(Animal):
    def fly(self):
        print("bird is flying")

# ab isko call krenge humnne Main class animal banaya or usko inherit kiya sare animal me to inherit krne se jo bhi animal ka function hai wo sare animal me apply hoga. 

bird1 = Bird()
dog1 = Dog()
fish1 = Fish()
bird1.eat()
bird1.fly()
dog1.eat()
dog1.bark()
fish1.eat()
fish1.swim()
```

Types of inheritance.

```python
# 1.single inheritance
# 2.multiple inheritance
# 3.multi level inhritance
# 4.hierarchical inheritance
# 5.hybrid inheritance
```

1. single inheritannce :-

```python
class Company:
    def location(self):
        print("location : Banglore")

class Employee(Company):
    def work(self):
        print("work : full stack developer")

# isme single class ko inherit kar rha hai isiliye single  inheritance kaha jata hai.isme Employee class inherit kar rha hai Comapny class ko.

emp1 = Employee()
emp1.work()
emp1.location()
```

2. multiple inheritance :-

```python
# isme multiple class ko ik hi class me inherit kiya jaa skta hai. 

#  father + mother = child

# Eg :-

class Father:
    def drive(self):
        print("Drive :")

class Mother:
    def cook(self):
        print("cook :")

class Child(Father, Mother):
    pass

ch1 = Child()
ch1.drive()
ch1.cook()
```

3. multi level inheritance :-

```python
# isme ik class dusre ko dusra tisre ko or ishi tarah multi leveling class banta hai usko multi level inheritance khte hai. 

# grandfather se father = father se child (isme child ke pass grandfather or father ka class ho jata hai :)

# Eg:-

class Granfather:
    def house(self):
        print("have house ;")

class Father(Granfather):
    def car(self):
        print("have car ;")

class Child(Father):
    pass

ch1 = Child()
ch1.car()
ch1.house()
```

4. heirarchical inhritance :-

```python
# isme single  parent multiple child  ko inherit kr skte hai. 

# Eg :-

class Animal:
    def eat(self):
        print("eating")

class Fish(Animal):       # ish tarah Animal class ko sabme inherit krenge jitne bhi class banenge sabme
    def swim(self):
        print("fish is swimming") 

class Dog(Animal):
    def bark(self):
        print("dog is barking")

class Bird(Animal):
    def fly(self):
        print("bird is flying")

# isme Animal class kayi Animal ko inherit krne de rha hai. 

bird1 = Bird()
dog1 = Dog()
fish1 = Fish()
bird1.eat()
bird1.fly()
dog1.eat()
dog1.bark()
fish1.eat()
fish1.swim()
```

5. hybrid inheritance :-

```python
# yaha pr upr ke 4 inheritance me se kisi ka bhi use ho skta hai ya sab mix ho skta hai. 

# Eg :-

class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(C):
    pass

class E(D):
    pass

class F(B):
    pass
```

MRO (method resolution order):-

```python
# MRO (method resolution order) is the order in which python searches for a method when a method is called on object.

# used when :
    # multiple inheritance
    # hybrid inheritance
    # with super()

# python uses :
    # C3 linearization algorithm

# Ensure :
    # left to right order
    # no parent is skipped
    # consistent heirachy

# Eg :-

class Father:
    def drive(self):
        print("father drive")

class Mother:
    def drive(self):
        print("mother drive")

class Child(Mother, Father):
    pass

ch1 = Child()
ch1.drive()

# isme father bhi drive kar skte hai or mother bhi to child kisko print krega jo phle inheri kiya ho child usko print krega.
```
