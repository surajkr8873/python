What is type conversion ?

```python
# English:

# Type conversion means changing one data type into another data type.

# Hindi:

# Type conversion ka matlab hai ek data type ki value ko doosre data type mein convert/change karna.
```

```python
a = "10" # >> "10" >string

print(type(a)) # >> 10 jo value hai wo integer valuee hai lekin string ke andar hai,isiliye a ka datatype string aayega.ishi data type ko change karne ko khte hai type conversion.
```

```python
# Eg:-
a = "25" # >> ye ik string data type hai ab isko convert karna hai.integer datatype me.
a = int(a) # >> yaha pr hum a ko integer value me change krne ka syntax de diye.
print(type(a)) # >> yaha pr hum a ka type check krenge ye type change hua ya nahi.
```

kisi string ko float datatype me change krna ho to.

```python
m = "25"
m = float(m)
print(type(m))
```

ya kisi float ya int ko string me change krna ho.

```python
z = 50.2
y = 25

z = str(z)
y = str(y)

print(type(z))
print(type(y))
```

```python
z = 50.2

z = bool(z)
print(type(z))
```

```python
z = 50.2
z = complex(z)
print(type(z))
```
