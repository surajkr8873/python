What is implicit  type conversion ?

```python
# English:

# Implicit conversion means Python automatically changes one data type into another when needed.

# Hindi:

# Implicit conversion ka matlab hai Python khud hi data type ko automatically change kar deta hai. Hume manually int() ya float() lagane ki zarurat nahi hoti.
```

```python
a = 5 # >> int >
b = 10.5 # >> float >
c = (a + b) # >> jab dono ko add krenge to float value niklega.

print(type(c)) # >> isme hum manually type nahi kr rhe hai python khud hi bata deta hai kki kis datatype me output hai.

#  output >> float datatype.> 
```

```python
# >> Eg:-1

a = 10
b = 5.5

c = a + b

print(c)
print(type(c))
```

```python
# >> Eg:-2

a = 10
b = 2 + 3j

c = a + b

print(c)
print(type(c))
```

```python
# >> Eg:-3

a = 5.5
b = 2 + 3j

c = a + b

print(c)
print(type(c))
```
