What is expllicit conversion ?

```python
# English:

# Explicit conversion means we manually change one data type into another datatype.

# Hindi:

# Explicit conversion ka matlab hai hum khud ek data type ko doosre data type mein convert karte hain
```

```python
# >> iske liye function use hota hai
int()
float()
str()
bool()
```

```python
# Eg:-1 >> string to integer

a = "10"

a = int(a)

print(a)
print(type(a))
```

```python
# Eg:-2 >> int to float.

a = 10
a = float(a)

print(a)
print(type(a))
```

```python
# Eg:-3 >> int to string.
a = 100
a = str(a)

print(type(a))
```

```python
# Eg:-4 >> int to bool.

a = 10
b = bool(a)

print(b)
```

```python
# Eg:-5 >> int to complex

a = 10
b = complex(a)

print(b)
```

```python
# Eg:-6 >> float to int.

a = 10.5
b = int(a)

print(b)
```

```python
# Eg:-7 >> bool to int.

a = True
b = int(a)

print(b)
```

```python
# Eg:-8 >> bool to int.

a = False

b = int(a)

print(b)
```
