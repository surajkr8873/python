What is string method ?

```python
# English

# String methods are built-in functions used to work with and modify strings.

# Hindi 

# String methods ka use string ke saath kaam karne ke liye hota hai, jaise uppercase karna, lowercase karna, spaces hatana, replace karna etc.
```

```python
name = "suraj"

# string "suraj" hamara data hai
# Ab hum uske saath kaam karne ke liye methods use karenge:

name.upper() # >> ishse sare letters capital ho jayega
```

5 useful string method

```python
name = "suraj yadav"

# 1. upper() → sab CAPITAL

print(name.upper()) # >> sare letter capital
# SURAJ YADAV

# 2. lower() → sab small

print(name.lower()) # >> sare letter small
# suraj yadav

# 3. title() → har word ka first letter CAPITAL

print(name.title()) # >> first letter capital
# Suraj Yadav

# 4. replace() → ek text ko doosre text se replace

print(name.replace("suraj", "rahul")) # >> text change
# rahul yadav

# 5. count() → koi character/word kitni baar hai

print(name.count("a")) # >> ik hi letter pure value me kitni baar aaya jo
# 2
```

```python
# Method	                   Kaam

# upper()                	   CAPITAL
# lower()	                   small
# title()	                   First Letter Capital
# replace()	                   text change
# count()	                   kitni baar hai
```
