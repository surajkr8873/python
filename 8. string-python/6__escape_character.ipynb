5 Important Escape Characters

```python
# 1. \n → New Line

print("Hello\nSuraj")
# Hello
# Suraj


# 2. \t → Tab / Space

print("Hello\tSuraj")
# Hello    Suraj


# 3. \" → Double quote print karna

print("He said \"Hello\"")
# He said "Hello"


# 4. \' → Single quote print karna

print('It\'s Python')
# It's Python


# 5. \\ → Backslash print karna

print("C:\\Users\\Suraj")
# C:\Users\Suraj
```

```python
# scape              	Kaam

# \n	                New line
# \t	                Tab/space
# \"	                Double quote
# \'	                Single quote
# \\	                Backslash   >>> phla (/) dusre (/) ko bolta hai "Tum special command nahi ho, mujhe literally ek backslash print karna hai."
```
