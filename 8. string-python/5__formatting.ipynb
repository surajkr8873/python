What is formatting in string?

```python
# English

# String formatting means putting variables/values inside a string in a proper way.

# Hindi

# String formatting ka matlab hai string ke andar variable ki value ko properly add karna.
```

```python
name = "Suraj"
age = 18
```

hume output chahiye

```python
# My name is Suraj and I am 18 years old.
```

```python
#  this is the code .

name = "Suraj"
age = 18

print(f"My name is {name} and I am {age} years old.")

# f → Python ko batata hai ki string ke andar variables hain.
# {name} → yahan name ki value aa jayegi. >>> sign {}
# {age} → yahan age ki value aa jayegi.
```

5 exampple of formatting string

```python
name = "Suraj"
age = 18
city = "Patna"

print(f"My name is {name}.")
# My name is Suraj.

print(f"I am {age} years old.")
# I am 18 years old.

print(f"My name is {name} and I am {age}.")
# My name is Suraj and I am 18.

print(f"I live in {city}.")
# I live in Patna.

print(f"My name is {name}, I am {age} years old and I live in {city}.")
# My name is Suraj, I am 18 years old and I live in Patna.
```
