5 Example of  slicing oof string

```python
# 1. First 3 characters

name = "Suraj"

print(name[0:3])
# Output: Sur


# 2. Index 1 se 4 tak

name = "Suraj"

print(name[1:4])
# Output: ura


# 3. Starting se kisi position tak

name = "Python"

print(name[:3])
# Output: Pyt


# 4. Kisi position se end tak

name = "Python"

print(name[2:])
# Output: thon


# 5. Negative slicing

name = "Suraj"

print(name[-4:-1])
# Output: ura
```

Slicing mein Step bhi de sakte hain

```python
# string[start : stop : step]
```

```python
name = "Python"

# P   y   t   h   o   n
# 0   1   2   3   4   5

print(name[0:6:2]) # >>> isme  jo last colon me hai wo step  ko chhodte hue chalega >> 2 hai to do  step skip krega >> 3 rhta to 3 step skip .
```

5 Example of skip slicing

```python
name = "ABCDEFGH"

print(name[0:8:2])
# A C E G
# Har 1 character ko skip karke next lega

print(name[0:8:3])
# A D G
# 2 character skip karke next lega

print(name[1:8:2])
# B D F H
# B se start, har 1 character skip

print(name[::2])
# A C E G
# Puri string mein 1-1 character skip

print(name[::3])
# A D G
# Puri string mein 2-2 character skip


# step = 1  → koi skip nahi
# step = 2  → 1 character skip
# step = 3  → 2 characters skip
# step = 4  → 3 characters skip
```
