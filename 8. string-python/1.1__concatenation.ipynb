What is concatination of string?

```python
# English 

# String concatenation means joining two or more strings together.

# Hindi

# String concatenation ka matlab hai do ya do se zyada strings ko jodna.
```

basic Example

```python
# "Hello" + "Bhai"
#      ↓
# "HelloBhai"
```

EXAMPLE

```python
first_name = "Suraj"
last_name = "Yadav"

full_name = first_name + last_name

print(full_name) # >>> name ke bich me space dene ke liye hum string ke andar hum space ka use krenge.jo ki hum niche wala me kiye hai
```

```python
first_name = "Suraj"
last_name = " Yadav"

full_name = first_name + last_name

print(full_name)       # ya hum space ka use na krke.hum extra string add krenge jaise niche hua hai.
```

```python
first_name = "Suraj"
last_name = " Yadav"

full_name = first_name + "" + last_name

print(full_name)  
```

5 Example of concatenation of string.

```python
# 1. Do strings ko jodna

a = "Hello"
b = "World"

print(a + b)
# Output: HelloWorld


# 2. Space ke saath jodna

first = "Suraj"
last = "Yadav"

print(first + " " + last)
# Output: Suraj Yadav


# 3. Greeting banana

name = "Suraj"

print("Hello " + name)
# Output: Hello Suraj


# 4. Do words ko jodna

a = "Good"
b = "Morning"

print(a + " " + b)
# Output: Good Morning


# 5. Sentence banana

name = "Suraj"
city = "Patna"

print("My name is " + name + " and I live in " + city)
# Output: My name is Suraj and I live in Patna
```
