What is indexing in string ?

```python
# English

# String indexing means accessing a particular character from a string using its position (index).

# Hindi

# String indexing ka matlab hai string ke andar kisi particular character ko uski position ke through nikalna.
```

```python
name = "SURAJ"

# String :  S   U   R   A   J
# Index  :  0   1   2   3   4 # >> python me indexiing hmesa 0 se strt hoti hai..
```

basic example

```python
name = "Suraj"

print(name[0])
print(name[1])
print(name[2])
print(name[3])
print(name[4])
```

NEGATIVE INDEXING .

```python
# String :  S   u   r   a   j
# Positive: 0   1   2   3   4
# Negative:-5  -4  -3  -2  -1
```

```python
name = "Suraj"

print(name[-1]) # >> negative indexing me last character se negative count start hoti hai.
print(name[-2])
print(name[-3])
print(name[-4])
print(name[-5]) 
```
